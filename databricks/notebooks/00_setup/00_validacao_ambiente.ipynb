```python
dados = [
    ("Elétrico", 120),
    ("Hibrído", 80),
    ("Combustão", 800)
]

colunas = ["tipo_veiculo", "quantidade"]
df = spark.createDataFrame(dados, colunas)
display(df)
```

```python
spark.sql("""
SELECT
    current_catalog() AS catalogo_atual,
    current_schema() AS schema_atual
""").show()
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.validacao_delta_veiculos")
```

```python
%sql
SELECT *
FROM workspace.default.validacao_delta_veiculos;
```
